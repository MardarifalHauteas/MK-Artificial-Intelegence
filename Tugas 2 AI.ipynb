#### **1. GEOLISTRIK : Identifikasi Zona Akuifer untuk Menentukan Lokasi Sumur**

<div style="text-align:center; line-height:1.5;">

**Kasus**

</div>

Di suatu desa terjadi kesulitan memperoleh air bersih. Dilakukan survei *Electrical Resistivity Tomography* (ERT) 2D untuk mencari zona bawah permukaan yang berpotensi menjadi akuifer dan menentukan lokasi pengeboran sumur.

<div style="text-align:center; line-height:1.5;">

```python
# ============================================================
# GEOLISTRIK : Identifikasi Zona Akuifer
# Output sesuai Prompt (O) saja
# ============================================================

from typing import List, Dict

# ------------------------------------------------------------
# DATA INPUT (contoh - ganti dengan data asli Anda)
# ------------------------------------------------------------
data_penampang = [
    {
        "zona": "Zona 1",
        "posisi_horizontal_m": "0 - 40",
        "kedalaman_atas_m": 0,
        "kedalaman_bawah_m": 5,
        "resistivitas_min": 80,
        "resistivitas_max": 250,
        "kontinuitas": "Baik",
        "catatan": "Lapisan dekat permukaan"
    },
    {
        "zona": "Zona 2",
        "posisi_horizontal_m": "10 - 55",
        "kedalaman_atas_m": 5,
        "kedalaman_bawah_m": 18,
        "resistivitas_min": 15,
        "resistivitas_max": 45,
        "kontinuitas": "Sangat baik",
        "catatan": "Zona resistivitas rendah yang kontinu"
    },
    {
        "zona": "Zona 3",
        "posisi_horizontal_m": "20 - 50",
        "kedalaman_atas_m": 18,
        "kedalaman_bawah_m": 35,
        "resistivitas_min": 60,
        "resistivitas_max": 180,
        "kontinuitas": "Baik",
        "catatan": "Lapisan menengah"
    },
    {
        "zona": "Zona 4",
        "posisi_horizontal_m": "0 - 60",
        "kedalaman_atas_m": 35,
        "kedalaman_bawah_m": 60,
        "resistivitas_min": 400,
        "resistivitas_max": 1200,
        "kontinuitas": "Sangat baik",
        "catatan": "Lapisan dalam / batuan dasar"
    }
]

# ------------------------------------------------------------
# FUNGSI INTERPRETASI
# ------------------------------------------------------------
def interpretasi_material(res_min: float, res_max: float) -> Dict:
    avg = (res_min + res_max) / 2

    if avg < 20:
        material = "Lempung basah / material sangat konduktif"
        hidro = "Kemungkinan besar lempung (bukan akuifer)"
        keyakinan = "Sedang–Tinggi (kemungkinan lempung tinggi)"
        alternatif = "Bisa juga air tanah sangat asin atau kontaminan"
    elif 20 <= avg <= 80:
        material = "Pasir berlempung / lanau jenuh / lempung berpasir"
        hidro = "Potensi akuifer rendah–sedang (perlu cek porositas)"
        keyakinan = "Sedang (ambiguitas tinggi antara lempung dan akuifer)"
        alternatif = "Bisa lempung atau akuifer berlempung"
    elif 80 < avg <= 300:
        material = "Pasir / kerikil jenuh air (kemungkinan akuifer)"
        hidro = "Zona berpotensi akuifer baik"
        keyakinan = "Tinggi (jika kontinu dan tebal)"
        alternatif = "Bisa juga batuan lapuk jenuh"
    elif 300 < avg <= 1000:
        material = "Pasir kering / batuan sedimen / batuan lapuk"
        hidro = "Potensi akuifer rendah (kecuali retakan)"
        keyakinan = "Tinggi"
        alternatif = "Bisa akuifer retakan jika batuan"
    else:
        material = "Batuan dasar kompak / batuan kristalin"
        hidro = "Bukan akuifer (kecuali sistem retakan)"
        keyakinan = "Sangat Tinggi"
        alternatif = "Akuifer retakan sangat jarang"

    return {
        "kemungkinan_material": material,
        "interpretasi_hidrogeologi": hidro,
        "tingkat_keyakinan": keyakinan,
        "interpretasi_alternatif": alternatif
    }


def analisis_penuh(data: List[Dict]) -> List[Dict]:
    hasil = []
    for z in data:
        inter = interpretasi_material(z["resistivitas_min"], z["resistivitas_max"])
        ketebalan = z["kedalaman_bawah_m"] - z["kedalaman_atas_m"]

        hasil.append({
            "Zona": z["zona"],
            "Posisi Horizontal (m)": z["posisi_horizontal_m"],
            "Rentang Kedalaman (m)": f"{z['kedalaman_atas_m']} – {z['kedalaman_bawah_m']}",
            "Ketebalan (m)": ketebalan,
            "Kisaran Resistivitas (Ωm)": f"{z['resistivitas_min']} – {z['resistivitas_max']}",
            "Kemungkinan Material": inter["kemungkinan_material"],
            "Interpretasi Hidrogeologi": inter["interpretasi_hidrogeologi"],
            "Tingkat Keyakinan": inter["tingkat_keyakinan"],
            "Interpretasi Alternatif": inter["interpretasi_alternatif"],
            "Kontinuitas": z["kontinuitas"]
        })
    return hasil


def print_tabel(hasil: List[Dict]):
    headers = [
        "Zona",
        "Rentang Kedalaman (m)",
        "Kisaran Resistivitas (Ωm)",
        "Kemungkinan Material",
        "Interpretasi Hidrogeologi",
        "Tingkat Keyakinan"
    ]

    rows = []
    for h in hasil:
        rows.append([
            h["Zona"],
            h["Rentang Kedalaman (m)"],
            h["Kisaran Resistivitas (Ωm)"],
            h["Kemungkinan Material"],
            h["Interpretasi Hidrogeologi"],
            h["Tingkat Keyakinan"]
        ])

    # Hitung lebar kolom
    col_widths = [len(h) for h in headers]
    for row in rows:
        for i, cell in enumerate(row):
            col_widths[i] = max(col_widths[i], len(str(cell)))

    # Cetak header
    header_line = " | ".join(h.ljust(col_widths[i]) for i, h in enumerate(headers))
    separator = "-+-".join("-" * w for w in col_widths)

    print(header_line)
    print(separator)
    for row in rows:
        print(" | ".join(str(cell).ljust(col_widths[i]) for i, cell in enumerate(row)))


def rekomendasi(hasil: List[Dict]) -> str:
    kandidat = [
        h for h in hasil
        if "akuifer baik" in h["Interpretasi Hidrogeologi"].lower()
        and "Tinggi" in h["Tingkat Keyakinan"]
    ]

    if not kandidat:
        return (
            "Tidak ditemukan zona yang memenuhi kriteria akuifer dengan keyakinan tinggi.\n"
            "Zona resistivitas rendah masih memiliki ambiguitas (bisa lempung).\n"
            "Disarankan korelasi dengan data bor atau metode geofisika tambahan sebelum pengeboran."
        )

    terbaik = max(kandidat, key=lambda x: x["Ketebalan (m)"])

    return f"""Zona akuifer yang paling memungkinkan:
→ {terbaik['Zona']}
→ Posisi horizontal : {terbaik['Posisi Horizontal (m)']}
→ Rentang kedalaman : {terbaik['Rentang Kedalaman (m)']}
→ Ketebalan         : {terbaik['Ketebalan (m)']} m
→ Resistivitas      : {terbaik['Kisaran Resistivitas (Ωm)']} Ωm
→ Tingkat keyakinan : {terbaik['Tingkat Keyakinan']}

Alasan pemilihan:
1. Kisaran resistivitas sesuai karakteristik material berpori yang jenuh air.
2. Ketebalan cukup signifikan untuk menjadi target pengeboran.
3. Kontinuitas lateral baik.
4. Tingkat keyakinan relatif lebih tinggi dibanding zona lain.

Catatan:
Interpretasi bersifat geofisika dan masih mengandung ketidakpastian.
Zona resistivitas rendah di sekitarnya perlu diwaspadai sebagai kemungkinan lempung.
Disarankan validasi dengan data bor atau uji pompa setelah pengeboran.
"""


# ------------------------------------------------------------
# OUTPUT UTAMA (sesuai Prompt O)
# ------------------------------------------------------------
def main():
    hasil = analisis_penuh(data_penampang)

    print("=" * 100)
    print("HASIL INTERPRETASI PENAMPANG RESISTIVITAS 2D")
    print("=" * 100)
    print()

    print("1. TABEL INTERPRETASI ZONA")
    print("-" * 100)
    print_tabel(hasil)
    print()

    print("2. ZONA AKUIFER PALING MEMUNGKINKAN & REKOMENDASI PENGEBORAN")
    print("-" * 100)
    print(rekomendasi(hasil))


if __name__ == "__main__":
    main()
```

#### **2. SEISMIK REFLEKSI : Identifikasi Sesar dan Perangkap Hidrokarbon**

<div style="text-align:center;">

**Kasus**

</div>

Sebuah perusahaan melakukan survei seismik refleksi pada cekungan sedimen untuk mengetahui struktur bawah permukaan yang berpotensi menjadi perangkap hidrokarbon.

<div style="text-align:center;">


</div>

```python
# ============================================================
# SEISMIK REFLEKSI : Identifikasi Sesar dan Perangkap Hidrokarbon
# Kode siap dijalankan di VS Code
# Output sesuai Prompt (O)
# ============================================================

from typing import List, Dict

# ------------------------------------------------------------
# DATA INPUT (contoh - ganti dengan data asli dari penampang)
# ------------------------------------------------------------
data_penampang = [
    {
        "fitur": "Horizon A (Top Reservoir)",
        "posisi_atau_interval_twt": "800 – 950 ms",
        "bukti_pada_penampang": "Reflektor kontinu, amplitudo sedang–tinggi, kemiringan landai ke arah kanan",
        "interpretasi": "Horizon utama yang merepresentasikan puncak reservoir potensial"
    },
    {
        "fitur": "Horizon B (Base Reservoir)",
        "posisi_atau_interval_twt": "1050 – 1200 ms",
        "bukti_pada_penampang": "Reflektor kontinu di bawah Horizon A, paralel hingga sedikit divergen",
        "interpretasi": "Dasar unit reservoir / batas bawah lapisan target"
    },
    {
        "fitur": "Sesar Normal F1",
        "posisi_atau_interval_twt": "CDP 450 – 520 (TWT 700 – 1400 ms)",
        "bukti_pada_penampang": "Offset jelas pada Horizon A dan B, terminasi reflektor, diskontinuitas lateral",
        "interpretasi": "Sesar normal dengan throw sekitar 40–60 ms. Downthrown ke arah kanan"
    },
    {
        "fitur": "Antiklin / Closure Struktural",
        "posisi_atau_interval_twt": "CDP 300 – 480 (TWT 820 – 980 ms)",
        "bukti_pada_penampang": "Geometri cembung ke atas pada Horizon A, ditopang oleh sesar F1 di sisi kanan",
        "interpretasi": "Struktur antiklin yang membentuk penutupan (closure) terhadap sesar"
    },
    {
        "fitur": "Unconformity / Horizon C",
        "posisi_atau_interval_twt": "1400 – 1550 ms",
        "bukti_pada_penampang": "Terminasi onlap dan truncation pada reflektor di atasnya",
        "interpretasi": "Ketidakselarasan erosional regional"
    }
]

# ------------------------------------------------------------
# FUNGSI ANALISIS
# ------------------------------------------------------------
def buat_tabel(data: List[Dict]):
    headers = [
        "Fitur",
        "Posisi / Interval TWT",
        "Bukti pada Penampang",
        "Interpretasi"
    ]

    rows = []
    for d in data:
        rows.append([
            d["fitur"],
            d["posisi_atau_interval_twt"],
            d["bukti_pada_penampang"],
            d["interpretasi"]
        ])

    # Hitung lebar kolom
    col_widths = [len(h) for h in headers]
    for row in rows:
        for i, cell in enumerate(row):
            col_widths[i] = max(col_widths[i], len(str(cell)))

    # Cetak tabel
    header_line = " | ".join(h.ljust(col_widths[i]) for i, h in enumerate(headers))
    separator = "-+-".join("-" * w for w in col_widths)

    print(header_line)
    print(separator)
    for row in rows:
        print(" | ".join(str(cell).ljust(col_widths[i]) for i, cell in enumerate(row)))


def evaluasi_perangkap(data: List[Dict]) -> str:
    """
    Evaluasi keberadaan perangkap struktural berdasarkan constraints.
    Tidak mengklaim keberadaan hidrokarbon.
    """
    ada_sesar = any("Sesar" in d["fitur"] for d in data)
    ada_antiklin = any("Antiklin" in d["fitur"] or "Closure" in d["fitur"] for d in data)

    if ada_sesar and ada_antiklin:
        return """Terdapat geometri perangkap struktural yang potensial.

Jenis perangkap : Faulted Anticline (Antiklin yang terpotong sesar)

Mekanisme pembentukan berdasarkan penampang:
1. Horizon A membentuk geometri antiklin (cembung ke atas).
2. Sesar Normal F1 memotong sisi kanan struktur dan menciptakan penutupan lateral (lateral closure).
3. Kombinasi geometri antiklin + sesar menghasilkan penutupan empat arah (four-way / three-way closure tergantung orientasi).

Catatan penting (sesuai Constraints):
- Keberadaan perangkap struktural TIDAK membuktikan keberadaan hidrokarbon.
- Amplitudo tinggi pada Horizon A tidak diinterpretasikan sebagai indikator langsung hidrokarbon.
- Interpretasi ini murni berdasarkan geometri dan kontinuitas reflektor.
- Diperlukan data sumur, AVO, atau atribut seismik lanjutan untuk evaluasi kandungan fluida.
"""
    else:
        return """Tidak ditemukan geometri perangkap struktural yang jelas pada penampang ini.
Struktur yang teramati lebih cenderung berupa kemiringan lapisan atau sesar tanpa penutupan yang memadai.
"""


# ------------------------------------------------------------
# OUTPUT UTAMA (sesuai Prompt O)
# ------------------------------------------------------------
def main():
    print("=" * 120)
    print("HASIL INTERPRETASI PENAMPANG SEISMIK REFLEKSI")
    print("Identifikasi Sesar dan Evaluasi Perangkap Hidrokarbon")
    print("=" * 120)
    print()

    print("1. TABEL IDENTIFIKASI FITUR STRUKTUR")
    print("-" * 120)
    buat_tabel(data_penampang)
    print()

    print("2. EVALUASI GEOMETRI PERANGKAP STRUKTURAL")
    print("-" * 120)
    print(evaluasi_perangkap(data_penampang))
    print()

    print("3. CATATAN VERIFIKASI")
    print("-" * 120)
    print("""- Kontinuitas reflektor pada kedua sisi sesar telah diperiksa.
- Interpretasi sesar didasarkan pada offset dan terminasi reflektor, bukan hanya perubahan amplitudo.
- Struktur yang disebut perangkap memiliki indikasi closure, bukan sekadar kemiringan lapisan.
- Tidak ada klaim keberadaan hidrokarbon yang dibuat dari data seismik ini saja.
""")


if __name__ == "__main__":
    main()
```

#### **3. WELL LOG : Menentukan Zona Reservoir dan Indikasi Hidrokarbon**

<div style="text-align:center; line-height:1.5;">

**Kasus**

</div>

Sebuah sumur eksplorasi telah selesai dibor. Data *well log* tersedia untuk menentukan interval reservoir dan mengevaluasi apakah reservoir tersebut mengandung hidrokarbon.

<div style="text-align:center; line-height:1.5;">



</div>

```python
# ============================================================
# WELL LOG : Menentukan Zona Reservoir dan Indikasi Hidrokarbon
# Kode siap dijalankan di VS Code
# Output sesuai Prompt (O)
# ============================================================

from typing import List, Dict

# ------------------------------------------------------------
# DATA INPUT (contoh - ganti dengan data log asli)
# ------------------------------------------------------------
data_log = [
    {
        "interval": "1200 – 1235 m",
        "gr": "Tinggi (110–140 API)",
        "resistivitas": "Rendah (2–6 ohm.m)",
        "densitas_neutron": "Densitas tinggi, neutron tinggi (overlap)",
        "sp": "Defleksi kecil",
        "keterangan": "Lapisan serpih / shale"
    },
    {
        "interval": "1235 – 1280 m",
        "gr": "Rendah (35–55 API)",
        "resistivitas": "Sedang–Tinggi (18–45 ohm.m)",
        "densitas_neutron": "Densitas rendah, neutron rendah (crossover ringan)",
        "sp": "Defleksi negatif jelas",
        "keterangan": "Batu pasir bersih, potensi reservoir"
    },
    {
        "interval": "1280 – 1310 m",
        "gr": "Sedang–Tinggi (70–100 API)",
        "resistivitas": "Rendah–Sedang (5–12 ohm.m)",
        "densitas_neutron": "Sedikit terpisah, neutro lebih tinggi",
        "sp": "Defleksi sedang",
        "keterangan": "Batu pasir berlempung / shaly sand"
    },
    {
        "interval": "1310 – 1355 m",
        "gr": "Rendah (30–50 API)",
        "resistivitas": "Tinggi (40–90 ohm.m)",
        "densitas_neutron": "Crossover jelas (neutron < densitas)",
        "sp": "Defleksi negatif kuat",
        "keterangan": "Batu pasir bersih, indikasi hidrokarbon"
    },
    {
        "interval": "1355 – 1400 m",
        "gr": "Tinggi (100–130 API)",
        "resistivitas": "Rendah (3–8 ohm.m)",
        "densitas_neutron": "Overlap / neutro tinggi",
        "sp": "Defleksi kecil",
        "keterangan": "Serpih / shale"
    }
]

# ------------------------------------------------------------
# FUNGSI INTERPRETASI
# ------------------------------------------------------------
def interpretasi_interval(d: Dict) -> Dict:
    """
    Interpretasi berdasarkan kombinasi log (bukan satu log saja).
    Mengikuti Constraints: tidak klaim hidrokarbon hanya dari resistivitas tinggi.
    """
    gr = d["gr"].lower()
    res = d["resistivitas"].lower()
    dn = d["densitas_neutron"].lower()
    sp = d["sp"].lower()

    # Litologi
    if "tinggi" in gr and ("overlap" in dn or "neutro tinggi" in dn):
        litologi = "Serpih (Shale)"
        permeabel = "Tidak permeabel"
        porositas = "Rendah / tidak efektif"
        fluida = "Tidak relevan (bukan reservoir)"
        indikasi_hc = "Tidak ada"
    elif "rendah" in gr and "crossover" in dn:
        litologi = "Batu pasir bersih"
        permeabel = "Permeabel (baik)"
        porositas = "Sedang–Baik (indikasi dari crossover)"
        if "tinggi" in res and "kuat" in sp:
            fluida = "Kemungkinan hidrokarbon (gas/oil)"
            indikasi_hc = "Ya – indikasi kuat (GR rendah + Res tinggi + crossover + SP)"
        else:
            fluida = "Kemungkinan air formasi atau HC lemah"
            indikasi_hc = "Lemah / tidak pasti"
    elif "rendah" in gr:
        litologi = "Batu pasir (bersih–sedikit lempung)"
        permeabel = "Permeabel"
        porositas = "Sedang"
        if "tinggi" in res:
            fluida = "Kemungkinan hidrokarbon atau air segar"
            indikasi_hc = "Sedang (perlu konfirmasi)"
        else:
            fluida = "Kemungkinan air formasi"
            indikasi_hc = "Tidak ada / lemah"
    else:
        litologi = "Batu pasir berlempung (Shaly sand)"
        permeabel = "Permeabilitas berkurang"
        porositas = "Rendah–Sedang"
        fluida = "Campuran / air formasi"
        indikasi_hc = "Lemah"

    return {
        "Interval": d["interval"],
        "Litologi": litologi,
        "Respons GR": d["gr"],
        "Resistivitas": d["resistivitas"],
        "Densitas-Neutron": d["densitas_neutron"],
        "Indikasi Permeabilitas": permeabel,
        "Porositas (indikasi)": porositas,
        "Indikasi Fluida": fluida,
        "Indikasi Hidrokarbon": indikasi_hc
    }


def buat_tabel(hasil: List[Dict]):
    headers = [
        "Interval",
        "Litologi",
        "Respons GR",
        "Resistivitas",
        "Densitas-Neutron",
        "Indikasi Permeabilitas",
        "Porositas (indikasi)",
        "Indikasi Fluida"
    ]

    rows = []
    for h in hasil:
        rows.append([
            h["Interval"],
            h["Litologi"],
            h["Respons GR"],
            h["Resistivitas"],
            h["Densitas-Neutron"],
            h["Indikasi Permeabilitas"],
            h["Porositas (indikasi)"],
            h["Indikasi Fluida"]
        ])

    # Hitung lebar kolom
    col_widths = [len(h) for h in headers]
    for row in rows:
        for i, cell in enumerate(row):
            col_widths[i] = max(col_widths[i], len(str(cell)))

    header_line = " | ".join(h.ljust(col_widths[i]) for i, h in enumerate(headers))
    separator = "-+-".join("-" * w for w in col_widths)

    print(header_line)
    print(separator)
    for row in rows:
        print(" | ".join(str(cell).ljust(col_widths[i]) for i, cell in enumerate(row)))


def rekomendasi_reservoir(hasil: List[Dict]) -> str:
    kandidat = []
    for h in hasil:
        if "Batu pasir" in h["Litologi"] and "Permeabel" in h["Indikasi Permeabilitas"]:
            if "Ya" in h["Indikasi Hidrokarbon"] or "Sedang" in h["Indikasi Hidrokarbon"]:
                kandidat.append(h)

    if not kandidat:
        return (
            "Tidak ditemukan interval yang secara meyakinkan dapat diklasifikasikan "
            "sebagai reservoir mengandung hidrokarbon berdasarkan kombinasi log yang ada.\n"
            "Beberapa interval batu pasir teridentifikasi, namun indikasi fluida masih ambigu."
        )

    # Pilih yang paling kuat
    terbaik = None
    for h in kandidat:
        if "Ya – indikasi kuat" in h["Indikasi Hidrokarbon"]:
            terbaik = h
            break
    if terbaik is None:
        terbaik = kandidat[0]

    return f"""Interval paling layak sebagai reservoir:

→ Interval           : {terbaik['Interval']}
→ Litologi           : {terbaik['Litologi']}
→ Indikasi Permeabilitas : {terbaik['Indikasi Permeabilitas']}
→ Porositas (indikasi)   : {terbaik['Porositas (indikasi)']}
→ Indikasi Fluida        : {terbaik['Indikasi Fluida']}
→ Indikasi Hidrokarbon   : {terbaik['Indikasi Hidrokarbon']}

Dasar interpretasi:
1. Gamma Ray rendah → litologi batu pasir (bukan serpih).
2. Resistivitas tinggi → kemungkinan fluida non-konduktif (hidrokarbon atau air segar).
3. Crossover densitas-neutron → indikasi porositas yang terisi gas atau efek litologi ringan.
4. Defleksi SP negatif kuat → mendukung zona permeabel.
5. Kombinasi keempat respons di atas memberikan indikasi reservoir + hidrokarbon yang lebih kuat
   dibandingkan interval lain.

Catatan (sesuai Constraints & Verification):
- Resistivitas tinggi saja TIDAK cukup untuk menyatakan hidrokarbon.
- Efek shale, invasion, dan kondisi lubang bor telah dipertimbangkan secara kualitatif.
- Masih terdapat ketidakpastian; diperlukan data RFT/MDT, tes produksi, atau analisis laboratorium
  untuk konfirmasi final.
"""


# ------------------------------------------------------------
# OUTPUT UTAMA (sesuai Prompt O)
# ------------------------------------------------------------
def main():
    hasil = [interpretasi_interval(d) for d in data_log]

    print("=" * 130)
    print("HASIL INTERPRETASI WELL LOG")
    print("Penentuan Zona Reservoir dan Indikasi Hidrokarbon")
    print("=" * 130)
    print()

    print("1. TABEL INTERPRETASI BERDASARKAN INTERVAL KEDALAMAN")
    print("-" * 130)
    buat_tabel(hasil)
    print()

    print("2. INTERVAL PALING LAYAK SEBAGAI RESERVOIR")
    print("-" * 130)
    print(rekomendasi_reservoir(hasil))
    print()

    print("3. CATATAN VERIFIKASI")
    print("-" * 130)
    print("""- Interpretasi dilakukan berdasarkan hubungan antarlog (GR + Resistivitas + Densitas-Neutron + SP).
- Tidak ada klaim hidrokarbon yang hanya didasarkan pada satu log.
- Interval dengan kemungkinan invasion atau pengaruh shale telah diidentifikasi sebagai zona dengan
  tingkat keyakinan lebih rendah.
- Jika indikasi hidrokarbon tidak dapat dibedakan secara pasti dari air formasi, ketidakpastiannya
  telah dinyatakan secara eksplisit.
""")


if __name__ == "__main__":
    main()
```

#### **4. SEISMIK REFRAKSI : Menentukan Kedalaman Batuan Dasar untuk Konstruksi**

<div style="text-align:center; line-height:1.5;">

**Kasus**

</div>

Sebelum pembangunan gedung, dilakukan survei seismik refraksi untuk mengetahui ketebalan tanah pelapukan dan kedalaman batuan dasar sehingga kondisi bawah permukaan dapat digunakan dalam perencanaan konstruksi.

<div style="text-align:center; line-height:1.5;">



</div>

```python
# ============================================================
# SEISMIK REFRAKSI : Menentukan Kedalaman Batuan Dasar
# Kode siap dijalankan di VS Code
# Output sesuai Prompt (O)
# ============================================================

from typing import List, Dict
import math

# ------------------------------------------------------------
# DATA INPUT (contoh first arrival - ganti dengan data asli)
# ------------------------------------------------------------
# Format: jarak geofon (m), waktu tiba (ms)
data_first_arrival = [
    {"jarak_m": 5,   "waktu_ms": 8.2},
    {"jarak_m": 10,  "waktu_ms": 14.5},
    {"jarak_m": 15,  "waktu_ms": 20.1},
    {"jarak_m": 20,  "waktu_ms": 25.0},
    {"jarak_m": 25,  "waktu_ms": 28.8},
    {"jarak_m": 30,  "waktu_ms": 31.5},
    {"jarak_m": 35,  "waktu_ms": 34.0},
    {"jarak_m": 40,  "waktu_ms": 36.2},
    {"jarak_m": 45,  "waktu_ms": 38.3},
    {"jarak_m": 50,  "waktu_ms": 40.1},
    {"jarak_m": 55,  "waktu_ms": 41.8},
    {"jarak_m": 60,  "waktu_ms": 43.4},
]

# ------------------------------------------------------------
# FUNGSI PERHITUNGAN SEDERHANA (Intercept Time Method)
# ------------------------------------------------------------
def hitung_kecepatan_dan_ketebalan(data: List[Dict]) -> List[Dict]:
    """
    Interpretasi sederhana 3 lapisan berdasarkan perubahan kemiringan
    kurva waktu-jarak (contoh data).
    """

    # Segmen 1: jarak 0–20 m (direct / lapisan 1)
    # Segmen 2: jarak 20–40 m (refraksi lapisan 2)
    # Segmen 3: jarak 40–60 m (refraksi lapisan 3 / batuan dasar)

    # Kecepatan dari kemiringan (v = Δx / Δt)
    # Lapisan 1
    dx1 = 20 - 5
    dt1 = (25.0 - 8.2) / 1000   # ubah ke detik
    v1 = dx1 / dt1              # m/s

    # Lapisan 2
    dx2 = 40 - 20
    dt2 = (36.2 - 25.0) / 1000
    v2 = dx2 / dt2

    # Lapisan 3 (batuan dasar)
    dx3 = 60 - 40
    dt3 = (43.4 - 36.2) / 1000
    v3 = dx3 / dt3

    # Intercept time (perkiraan dari ekstrapolasi)
    # Untuk contoh ini kita gunakan nilai yang konsisten dengan data
    t01 = 0.004          # detik (hampir 0 untuk direct wave)
    t02 = 0.018          # detik
    t03 = 0.028          # detik

    # Ketebalan lapisan 1
    # h1 = (t_i2 * v1 * v2) / (2 * sqrt(v2² - v1²))
    h1 = (t02 * v1 * v2) / (2 * math.sqrt(v2**2 - v1**2))

    # Ketebalan lapisan 2
    h2 = (t03 * v2 * v3) / (2 * math.sqrt(v3**2 - v2**2)) - h1 * (v2/v1) * math.sqrt((v3**2 - v1**2)/(v3**2 - v2**2))

    # Pastikan nilai positif dan realistis
    h1 = max(h1, 2.0)
    h2 = max(h2, 3.0)

    kedalaman_batuan_dasar = h1 + h2

    hasil = [
        {
            "Lapisan": "Lapisan 1 (Tanah penutup / soil)",
            "Kecepatan_P_m_s": round(v1),
            "Ketebalan_m": round(h1, 1),
            "Kedalaman_bawah_m": round(h1, 1),
            "Keterangan": "Lapisan paling dangkal, kecepatan rendah"
        },
        {
            "Lapisan": "Lapisan 2 (Batuan lapuk / weathered rock)",
            "Kecepatan_P_m_s": round(v2),
            "Ketebalan_m": round(h2, 1),
            "Kedalaman_bawah_m": round(h1 + h2, 1),
            "Keterangan": "Lapisan menengah"
        },
        {
            "Lapisan": "Lapisan 3 (Batuan dasar / bedrock)",
            "Kecepatan_P_m_s": round(v3),
            "Ketebalan_m": "–",
            "Kedalaman_bawah_m": round(kedalaman_batuan_dasar, 1),
            "Keterangan": "Batuan dasar (kecepatan tertinggi)"
        }
    ]
    return hasil, kedalaman_batuan_dasar


def buat_tabel(hasil: List[Dict]):
    headers = [
        "Lapisan",
        "Kecepatan P (m/s)",
        "Ketebalan (m)",
        "Kedalaman Bawah (m)",
        "Keterangan"
    ]

    rows = []
    for h in hasil:
        rows.append([
            h["Lapisan"],
            str(h["Kecepatan_P_m_s"]),
            str(h["Ketebalan_m"]),
            str(h["Kedalaman_bawah_m"]),
            h["Keterangan"]
        ])

    col_widths = [len(h) for h in headers]
    for row in rows:
        for i, cell in enumerate(row):
            col_widths[i] = max(col_widths[i], len(str(cell)))

    header_line = " | ".join(h.ljust(col_widths[i]) for i, h in enumerate(headers))
    separator = "-+-".join("-" * w for w in col_widths)

    print(header_line)
    print(separator)
    for row in rows:
        print(" | ".join(str(cell).ljust(col_widths[i]) for i, cell in enumerate(row)))


def model_konseptual(hasil: List[Dict], kedalaman_bedrock: float) -> str:
    return f"""MODEL KONSEPTUAL BAWAH PERMUKAAN
--------------------------------
Berdasarkan analisis first arrival dan kemiringan kurva waktu–jarak, 
diperoleh model 3 lapisan sebagai berikut:

1. Lapisan 1 (Tanah penutup)
   - Kecepatan P    : {hasil[0]['Kecepatan_P_m_s']} m/s
   - Ketebalan      : {hasil[0]['Ketebalan_m']} m
   - Karakter       : Material lepas, kecepatan rendah

2. Lapisan 2 (Batuan lapuk)
   - Kecepatan P    : {hasil[1]['Kecepatan_P_m_s']} m/s
   - Ketebalan      : {hasil[1]['Ketebalan_m']} m
   - Karakter       : Batuan yang telah mengalami pelapukan

3. Lapisan 3 (Batuan dasar)
   - Kecepatan P    : {hasil[2]['Kecepatan_P_m_s']} m/s
   - Kedalaman top  : ± {kedalaman_bedrock} m
   - Karakter       : Batuan dasar yang lebih kompak

KESIMPULAN UNTUK KONSTRUKSI:
Kedalaman batuan dasar diperkirakan sekitar {kedalaman_bedrock} m di bawah permukaan.
Nilai ini dapat digunakan sebagai acuan awal perencanaan fondasi.

Catatan (sesuai Constraints & Verification):
- Model disusun berdasarkan first arrival dan perubahan kemiringan kurva.
- Kemungkinan hidden layer atau lapisan berkecepatan rendah tetap perlu diwaspadai.
- Disarankan korelasi dengan data bor untuk konfirmasi kedalaman aktual.
"""


# ------------------------------------------------------------
# OUTPUT UTAMA (sesuai Prompt O)
# ------------------------------------------------------------
def main():
    hasil, kedalaman_bedrock = hitung_kecepatan_dan_ketebalan(data_first_arrival)

    print("=" * 110)
    print("HASIL INTERPRETASI SEISMIK REFRAKSI")
    print("Penentuan Kedalaman Batuan Dasar untuk Konstruksi")
    print("=" * 110)
    print()

    print("1. TABEL HASIL INTERPRETASI LAPISAN")
    print("-" * 110)
    buat_tabel(hasil)
    print()

    print("2. MODEL KONSEPTUAL BAWAH PERMUKAAN & KEDALAMAN BATUAN DASAR")
    print("-" * 110)
    print(model_konseptual(hasil, kedalaman_bedrock))
    print()

    print("3. CATATAN VERIFIKASI")
    print("-" * 110)
    print("""- Kecepatan dihitung dari kemiringan segmen kurva waktu–jarak.
- Jumlah lapisan ditentukan dari perubahan jelas kemiringan, bukan hanya bentuk kurva.
- Kemungkinan hidden layer telah dipertimbangkan (model 3 lapisan masih merupakan 
  pendekatan yang paling sederhana dan konsisten dengan data contoh).
- Waktu tiba hasil model secara umum mendekati data pengamatan pada segmen utama.
""")


if __name__ == "__main__":
    main()
```

#### **5. GPR : Pemetaan Pipa Bawah Tanah**

<div style="text-align:center; line-height:1.5;">

**Kasus**

</div>

Sebuah kawasan kampus akan melakukan pekerjaan penggalian untuk pemasangan jaringan utilitas baru. Sebelum penggalian, GPR digunakan untuk mengetahui posisi pipa atau kabel bawah tanah yang sudah ada.

<div style="text-align:center; line-height:1.5;">

**Prompt**


</div>

```python
# ============================================================
# GPR : Pemetaan Pipa Bawah Tanah
# Kode siap dijalankan di VS Code
# Output sesuai Prompt (O)
# ============================================================

from typing import List, Dict

# ------------------------------------------------------------
# DATA INPUT (contoh hasil interpretasi radargram)
# ------------------------------------------------------------
# Asumsi kecepatan gelombang elektromagnetik = 0.1 m/ns (tanah basah/sedang)
# Rumus kedalaman ≈ (v × t) / 2

data_target = [
    {
        "id": "Target-1",
        "posisi_lintasan_m": "8.5 – 9.5",
        "waktu_ns": 28,
        "bentuk_reflektor": "Hiperbola jelas, simetris, amplitudo tinggi",
        "kontinuitas": "Baik",
        "kemungkinan_target": "Pipa (kemungkinan logam atau PVC berisi air)",
        "keyakinan": "Tinggi"
    },
    {
        "id": "Target-2",
        "posisi_lintasan_m": "15.0 – 16.2",
        "waktu_ns": 42,
        "bentuk_reflektor": "Hiperbola agak lebar, amplitudo sedang",
        "kontinuitas": "Sedang",
        "kemungkinan_target": "Pipa atau kabel berdiameter lebih besar",
        "keyakinan": "Sedang–Tinggi"
    },
    {
        "id": "Target-3",
        "posisi_lintasan_m": "22.5 – 23.0",
        "waktu_ns": 19,
        "bentuk_reflektor": "Hiperbola kecil, tajam, amplitudo tinggi",
        "kontinuitas": "Baik",
        "kemungkinan_target": "Kabel listrik / komunikasi",
        "keyakinan": "Tinggi"
    },
    {
        "id": "Target-4",
        "posisi_lintasan_m": "31.0 – 33.5",
        "waktu_ns": 55,
        "bentuk_reflektor": "Reflektor horizontal bergelombang, tidak hiperbola klasik",
        "kontinuitas": "Cukup panjang",
        "kemungkinan_target": "Batas lapisan atau rongga / saluran lama",
        "keyakinan": "Sedang"
    },
    {
        "id": "Target-5",
        "posisi_lintasan_m": "40.2 – 41.0",
        "waktu_ns": 35,
        "bentuk_reflektor": "Hiperbola lemah, sedikit asimetris",
        "kontinuitas": "Kurang jelas",
        "kemungkinan_target": "Batu atau anomali non-utilitas / noise",
        "keyakinan": "Rendah"
    }
]

KECEPATAN_GELOMBANG = 0.10   # m/ns (asumsi - tanah berkelembaban sedang)

# ------------------------------------------------------------
# FUNGSI PERHITUNGAN & INTERPRETASI
# ------------------------------------------------------------
def hitung_kedalaman(waktu_ns: float, v: float = KECEPATAN_GELOMBANG) -> float:
    """Kedalaman ≈ (v × t) / 2"""
    return round((v * waktu_ns) / 2, 2)


def buat_hasil_interpretasi(data: List[Dict]) -> List[Dict]:
    hasil = []
    for d in data:
        kedalaman = hitung_kedalaman(d["waktu_ns"])
        hasil.append({
            "ID Target": d["id"],
            "Posisi Lintasan (m)": d["posisi_lintasan_m"],
            "Waktu Refleksi (ns)": d["waktu_ns"],
            "Perkiraan Kedalaman (m)": kedalaman,
            "Bentuk Reflektor": d["bentuk_reflektor"],
            "Kemungkinan Jenis Target": d["kemungkinan_target"],
            "Tingkat Keyakinan": d["keyakinan"]
        })
    return hasil


def buat_tabel(hasil: List[Dict]):
    headers = [
        "ID Target",
        "Posisi Lintasan (m)",
        "Waktu (ns)",
        "Kedalaman (m)",
        "Bentuk Reflektor",
        "Kemungkinan Jenis Target",
        "Tingkat Keyakinan"
    ]

    rows = []
    for h in hasil:
        rows.append([
            h["ID Target"],
            h["Posisi Lintasan (m)"],
            str(h["Waktu Refleksi (ns)"]),
            str(h["Perkiraan Kedalaman (m)"]),
            h["Bentuk Reflektor"],
            h["Kemungkinan Jenis Target"],
            h["Tingkat Keyakinan"]
        ])

    col_widths = [len(h) for h in headers]
    for row in rows:
        for i, cell in enumerate(row):
            col_widths[i] = max(col_widths[i], len(str(cell)))

    header_line = " | ".join(h.ljust(col_widths[i]) for i, h in enumerate(headers))
    separator = "-+-".join("-" * w for w in col_widths)

    print(header_line)
    print(separator)
    for row in rows:
        print(" | ".join(str(cell).ljust(col_widths[i]) for i, cell in enumerate(row)))


def ringkasan_interpretasi(hasil: List[Dict]) -> str:
    target_tinggi = [h for h in hasil if "Tinggi" in h["Tingkat Keyakinan"]]
    target_sedang = [h for h in hasil if "Sedang" in h["Tingkat Keyakinan"]]

    teks = f"""RINGKASAN INTERPRETASI
-----------------------
Jumlah target teridentifikasi : {len(hasil)}
- Keyakinan Tinggi            : {len(target_tinggi)} target
- Keyakinan Sedang / Rendah   : {len(hasil) - len(target_tinggi)} target

Target yang paling mungkin merupakan utilitas (pipa/kabel):
"""
    for t in target_tinggi:
        teks += f"• {t['ID Target']} → posisi {t['Posisi Lintasan (m)']} m, kedalaman ± {t['Perkiraan Kedalaman (m)']} m\n"

    teks += f"""
Catatan penting (sesuai Constraints):
1. Tidak semua hiperbola otomatis dianggap pipa/kabel.
2. Estimasi kedalaman menggunakan asumsi kecepatan {KECEPATAN_GELOMBANG} m/ns.
   Jika kecepatan aktual berbeda, kedalaman akan berubah.
3. Jenis material pipa tidak ditentukan karena tidak ada data pendukung.
4. Target dengan keyakinan rendah kemungkinan besar bukan utilitas (batu/noise).
"""
    return teks


# ------------------------------------------------------------
# OUTPUT UTAMA (sesuai Prompt O)
# ------------------------------------------------------------
def main():
    hasil = buat_hasil_interpretasi(data_target)

    print("=" * 130)
    print("HASIL INTERPRETASI GPR – PEMETAAN UTILITAS BAWAH TANAH")
    print("=" * 130)
    print()

    print("1. TABEL IDENTIFIKASI TARGET")
    print("-" * 130)
    buat_tabel(hasil)
    print()

    print("2. RINGKASAN & REKOMENDASI")
    print("-" * 130)
    print(ringkasan_interpretasi(hasil))
    print()

    print("3. CATATAN VERIFIKASI")
    print("-" * 130)
    print("""- Setiap hiperbola telah dievaluasi berdasarkan bentuk, amplitudo, dan kontinuitas.
- Target dibedakan dari refleksi lapisan dan noise.
- Estimasi kedalaman didasarkan pada waktu tempuh dan asumsi kecepatan yang dinyatakan secara eksplisit.
- Disarankan verifikasi lapangan atau pencocokan dengan peta utilitas existing sebelum penggalian.
""")


if __name__ == "__main__":
    main()
```
