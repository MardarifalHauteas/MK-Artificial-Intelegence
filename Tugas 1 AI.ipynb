#### **1. GEOLISTRIK : Identifikasi Zona Akuifer untuk Menentukan Lokasi Sumur**

<div style="text-align:center; line-height:1.5;">

**Kasus**

</div>

Di suatu desa terjadi kesulitan memperoleh air bersih. Dilakukan survei *Electrical Resistivity Tomography* (ERT) 2D untuk mencari zona bawah permukaan yang berpotensi menjadi akuifer dan menentukan lokasi pengeboran sumur.

<div style="text-align:center; line-height:1.5;">

**Prompt**

</div>

**Role (R):**  
Bertindak sebagai seorang geofisikawan hidrogeologi yang berpengalaman dalam interpretasi metode *Electrical Resistivity Tomography* (ERT) untuk investigasi air tanah.

**Task (T):**  
Analisis penampang resistivitas 2D untuk mengidentifikasi lapisan bawah permukaan, menentukan zona yang berpotensi mengandung air tanah, memperkirakan kedalaman dan ketebalan zona akuifer, serta memberikan rekomendasi lokasi pengeboran sumur berdasarkan hasil interpretasi.

**Context (C):**  
Survei dilakukan pada daerah yang mengalami keterbatasan air bersih. Kondisi geologi daerah tersusun atas lapisan tanah penutup, material sedimen, dan batuan dasar. Air tanah diperkirakan berada pada lapisan material yang memiliki porositas dan permeabilitas cukup baik. Namun, zona resistivitas rendah tidak selalu menunjukkan keberadaan air karena material lempung juga dapat memiliki resistivitas rendah.

**Input (I):**  
Gunakan penampang resistivitas 2D yang diberikan, termasuk posisi horizontal, kedalaman, nilai resistivitas, konfigurasi elektroda, spasi elektroda, dan informasi geologi daerah apabila tersedia.

**Constraints (K):**  
1. Jangan menentukan akuifer hanya berdasarkan satu nilai resistivitas.  
2. Pertimbangkan kemungkinan resistivitas rendah disebabkan oleh lempung, air tanah, atau material konduktif lainnya.  
3. Jangan menyatakan keberadaan air tanah sebagai kepastian jika tidak didukung data lain.  
4. Bedakan antara hasil pengukuran, interpretasi, dan asumsi.  
5. Jangan membuat data geologi atau nilai resistivitas yang tidak diberikan.

**Output (O):**  
Buat hasil interpretasi dalam bentuk tabel yang memuat zona, rentang kedalaman, kisaran resistivitas, kemungkinan material, interpretasi hidrogeologi, dan tingkat keyakinan. Setelah itu, jelaskan zona akuifer yang paling memungkinkan dan alasan pemilihannya. Jika data memungkinkan, tentukan titik atau rentang posisi yang paling sesuai untuk pengeboran.

**Quality Criteria (Q):**  
Interpretasi harus didasarkan pada pola resistivitas, kedalaman, kontinuitas zona, kondisi geologi, dan karakteristik material pembawa air. Interpretasi harus membedakan zona akuifer dari zona lempung atau material konduktif lainnya serta menyertakan tingkat ketidakpastian.

**Verification (V):**  
Periksa kembali apakah zona yang disebut sebagai akuifer memiliki karakteristik resistivitas dan geologi yang mendukung. Periksa juga kemungkinan interpretasi alternatif, terutama apabila zona resistivitas rendah dapat disebabkan oleh lempung. Jangan memberikan rekomendasi pengeboran jika bukti geofisikanya tidak cukup.

#### **2. SEISMIK REFLEKSI : Identifikasi Sesar dan Perangkap Hidrokarbon**

<div style="text-align:center;">

**Kasus**

</div>

Sebuah perusahaan melakukan survei seismik refleksi pada cekungan sedimen untuk mengetahui struktur bawah permukaan yang berpotensi menjadi perangkap hidrokarbon.

<div style="text-align:center;">

**Prompt**

</div>

**Role (R):**  
Bertindak sebagai seorang geofisikawan eksplorasi yang berpengalaman dalam pengolahan dan interpretasi *seismic reflection* untuk pemetaan struktur bawah permukaan pada cekungan sedimen.

**Task (T):**  
Interpretasikan penampang seismik refleksi untuk menentukan horizon utama, mengidentifikasi sesar, menentukan geometri dan kemiringan lapisan, serta mengevaluasi apakah terdapat struktur yang secara geometri dapat berperan sebagai perangkap hidrokarbon.

**Context (C):**  
Penampang seismik diperoleh dari suatu cekungan sedimen. Reflektor merepresentasikan perubahan sifat elastik bawah permukaan. Struktur seperti sesar, antiklin, dan *unconformity* dapat memengaruhi konfigurasi lapisan dan berpotensi membentuk perangkap. Keberadaan perangkap struktural tidak secara langsung membuktikan keberadaan hidrokarbon.

**Input (I):**  
Gunakan penampang seismik yang diberikan beserta skala TWT (*two-way travel time*), posisi CDP atau jarak, informasi kecepatan jika tersedia, serta data sumur atau informasi geologi regional jika tersedia.

**Constraints (K):**  
1. Jangan menginterpretasikan hidrokarbon hanya dari amplitudo reflektor.  
2. Jangan menyatakan suatu reflektor sebagai sesar hanya karena terdapat perubahan amplitudo.  
3. Gunakan kontinuitas, terminasi, *offset*, dan perubahan posisi reflektor sebagai dasar identifikasi struktur.  
4. Bedakan antara struktur yang benar-benar teridentifikasi dan interpretasi mengenai potensinya sebagai perangkap.  
5. Jangan membuat horizon, sesar, atau data kecepatan yang tidak terlihat atau tidak diberikan.

**Output (O):**  
Identifikasi dan jelaskan horizon utama, sesar, kemiringan lapisan, serta struktur geologi penting. Buat tabel yang berisi fitur, posisi atau interval TWT, bukti pada penampang, dan interpretasinya. Selanjutnya jelaskan apakah terdapat geometri perangkap struktural dan bagaimana mekanisme terbentuknya berdasarkan penampang.

**Quality Criteria (Q):**  
Interpretasi harus mengikuti kontinuitas dan terminasi reflektor serta memiliki bukti visual yang jelas. Interpretasi struktur harus konsisten secara geologi dan tidak boleh mengubah indikasi struktur menjadi klaim keberadaan hidrokarbon.

**Verification (V):**  
Periksa kembali kontinuitas reflektor pada kedua sisi struktur. Pastikan interpretasi sesar memiliki *offset* atau ketidakselarasan reflektor yang jelas. Periksa apakah struktur yang dianggap sebagai perangkap benar-benar memiliki konfigurasi penutupan (*closure*) atau hanya berupa kemiringan lapisan.

</div>

#### **3. WELL LOG : Menentukan Zona Reservoir dan Indikasi Hidrokarbon**

<div style="text-align:center; line-height:1.5;">

**Kasus**

</div>

Sebuah sumur eksplorasi telah selesai dibor. Data *well log* tersedia untuk menentukan interval reservoir dan mengevaluasi apakah reservoir tersebut mengandung hidrokarbon.

<div style="text-align:center; line-height:1.5;">

**Prompt**

</div>

**Role (R):**  
Bertindak sebagai seorang petrofisikawan dan geofisikawan perminyakan yang berpengalaman dalam interpretasi *well log* untuk karakterisasi litologi, porositas, saturasi fluida, dan evaluasi reservoir.

**Task (T):**  
Interpretasikan kombinasi log Gamma Ray, resistivitas, densitas, neutron, dan SP untuk menentukan litologi, zona permeabel, porositas, jenis fluida, serta interval yang memiliki indikasi hidrokarbon.

**Context (C):**  
Sumur menembus beberapa lapisan sedimen yang terdiri atas serpih dan batu pasir. Tujuan analisis adalah menentukan interval batu pasir yang memiliki karakteristik reservoir dan mengevaluasi kandungan fluidanya. Respons beberapa log dapat dipengaruhi oleh *shale*, *invasion*, kondisi lubang bor, dan sifat fluida formasi.

**Input (I):**  
Gunakan data log GR, SP, resistivitas, densitas, neutron, kedalaman, nilai *Rw* jika tersedia, serta informasi *mud properties* dan jenis formasi apabila diberikan.

**Constraints (K):**  
1. Jangan menentukan litologi berdasarkan satu log saja jika terdapat beberapa log.  
2. Jangan menyatakan suatu interval mengandung hidrokarbon hanya karena resistivitasnya tinggi.  
3. Pertimbangkan efek *shale*, *invasion*, *washout*, dan kondisi lubang bor.  
4. Bedakan antara indikasi reservoir dan indikasi hidrokarbon.  
5. Jangan membuat nilai *Rw*, porositas, saturasi air, atau parameter petrofisika lain yang tidak diberikan.

**Output (O):**  
Buat tabel interpretasi berdasarkan interval kedalaman yang berisi litologi, respons GR, resistivitas, densitas-neutron, indikasi permeabilitas, porositas jika dapat dihitung, serta indikasi fluida. Tentukan interval yang paling layak dianggap sebagai reservoir dan jelaskan dasar interpretasinya.

**Quality Criteria (Q):**  
Interpretasi harus menunjukkan hubungan antarlog, bukan hanya membaca satu kurva. Zona reservoir harus memiliki dukungan dari litologi, porositas, dan permeabilitas. Indikasi hidrokarbon harus didukung oleh kombinasi respons log yang relevan.

**Verification (V):**  
Periksa konsistensi antara GR, resistivitas, densitas, neutron, dan SP. Identifikasi interval yang mungkin mengalami *invasion* atau *washout*. Jika interpretasi hidrokarbon tidak dapat dibedakan secara pasti dari air formasi, nyatakan ketidakpastiannya.

#### **4. SEISMIK REFRAKSI : Menentukan Kedalaman Batuan Dasar untuk Konstruksi**

<div style="text-align:center; line-height:1.5;">

**Kasus**

</div>

Sebelum pembangunan gedung, dilakukan survei seismik refraksi untuk mengetahui ketebalan tanah pelapukan dan kedalaman batuan dasar sehingga kondisi bawah permukaan dapat digunakan dalam perencanaan konstruksi.

<div style="text-align:center; line-height:1.5;">

**Prompt**

</div>

**Role (R):**  
Bertindak sebagai seorang geofisikawan teknik yang berpengalaman dalam metode *seismic refraction* untuk investigasi kondisi bawah permukaan dangkal.

**Task (T):**  
Analisis data waktu tiba pertama (*first arrival*) untuk menentukan jumlah lapisan, kecepatan gelombang P setiap lapisan, ketebalan lapisan, serta kedalaman batuan dasar pada lokasi survei.

**Context (C):**  
Survei dilakukan pada lokasi yang akan digunakan untuk pembangunan gedung. Kondisi bawah permukaan diperkirakan terdiri atas tanah penutup, lapisan batuan lapuk, dan batuan dasar. Informasi kedalaman batuan dasar diperlukan untuk memahami kondisi geoteknik lokasi.

**Input (I):**  
Gunakan data posisi sumber, posisi geofon, jarak antar geofon, waktu tiba pertama, kurva waktu–jarak, dan informasi geologi permukaan jika tersedia.

**Constraints (K):**  
1. Gunakan *first arrival* sebagai dasar penentuan model.  
2. Tentukan kecepatan berdasarkan kemiringan segmen kurva waktu–jarak.  
3. Jangan menentukan jumlah lapisan hanya dari bentuk kurva tanpa analisis waktu tiba.  
4. Perhatikan kemungkinan *hidden layer*, lapisan berkecepatan rendah, dan perubahan lateral.  
5. Jangan menganggap lapisan dengan kecepatan tinggi secara otomatis sebagai batuan dasar tanpa mempertimbangkan konteks geologi.

**Output (O):**  
Tentukan jumlah lapisan, kecepatan gelombang P, ketebalan masing-masing lapisan, dan kedalaman batuan dasar. Sajikan hasil dalam tabel dan buat model konseptual bawah permukaan berdasarkan hasil interpretasi.

**Quality Criteria (Q):**  
Model harus mampu menjelaskan pola *travel-time curve*. Nilai kecepatan harus memiliki hubungan yang logis antar lapisan dan kedalaman hasil perhitungan harus konsisten dengan data lapangan.

**Verification (V):**  
Bandingkan waktu tiba hasil model dengan waktu tiba pengamatan. Identifikasi data yang memiliki penyimpangan besar. Periksa kemungkinan *hidden layer* atau lapisan berkecepatan rendah sebelum menyimpulkan jumlah lapisan dan kedalaman batuan dasar.

</div>

#### **5. GPR : Pemetaan Pipa Bawah Tanah**

<div style="text-align:center; line-height:1.5;">

**Kasus**

</div>

Sebuah kawasan kampus akan melakukan pekerjaan penggalian untuk pemasangan jaringan utilitas baru. Sebelum penggalian, GPR digunakan untuk mengetahui posisi pipa atau kabel bawah tanah yang sudah ada.

<div style="text-align:center; line-height:1.5;">

**Prompt**

</div>

**Role (R):**  
Bertindak sebagai seorang geofisikawan yang berpengalaman dalam metode *Ground Penetrating Radar* (GPR) untuk pemetaan utilitas bawah tanah.

**Task (T):**  
Interpretasikan radargram GPR untuk mengidentifikasi reflektor yang kemungkinan berasal dari pipa atau kabel bawah tanah, menentukan posisi horizontalnya, dan memperkirakan kedalamannya.

**Context (C):**  
Survei dilakukan di area kampus yang memiliki berbagai utilitas bawah tanah. Tujuan utama survei adalah memetakan posisi utilitas sebelum dilakukan penggalian. Radargram dapat mengandung refleksi dari pipa, kabel, batu, rongga, batas lapisan, maupun *noise*.

**Input (I):**  
Gunakan radargram, posisi dan panjang lintasan, frekuensi antena, skala waktu, jarak lintasan, serta nilai kecepatan gelombang elektromagnetik jika tersedia.

**Constraints (K):**  
1. Jangan langsung menganggap semua hiperbola sebagai pipa atau kabel.  
2. Gunakan bentuk, posisi, amplitudo, kontinuitas, dan karakteristik reflektor sebagai dasar interpretasi.  
3. Bedakan target utilitas dari refleksi lapisan dan *noise*.  
4. Jangan menentukan jenis material pipa tanpa informasi pendukung.  
5. Jika kecepatan gelombang tidak diketahui, jangan memberikan kedalaman pasti tanpa menjelaskan asumsi yang digunakan.

**Output (O):**  
Identifikasi setiap target yang berpotensi sebagai utilitas bawah tanah. Untuk setiap target, tentukan posisi lintasan, waktu kemunculan refleksi, perkiraan kedalaman, bentuk reflektor, kemungkinan jenis target, dan tingkat keyakinan interpretasi. Sajikan hasil dalam tabel.

**Quality Criteria (Q):**  
Target yang diinterpretasikan sebagai utilitas harus memiliki karakteristik refleksi yang mendukung, terutama pola hiperbola yang konsisten. Estimasi kedalaman harus berasal dari hubungan waktu tempuh dan kecepatan gelombang yang digunakan.

**Verification (V):**  
Periksa kembali setiap hiperbola dan bandingkan dengan pola reflektor di sekitarnya. Evaluasi kemungkinan bahwa target merupakan pipa, kabel, batu, rongga, atau *noise*. Jika tersedia, cocokkan hasil interpretasi dengan peta utilitas atau informasi lapangan.
